````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (2) 建立資料庫與初始資料
# 每次執行都先刪除舊表，確保資料庫回到初始狀態
def init_db():
    conn = sqlite3.connect('cafe.db')
    cursor = conn.cursor()

    # 刪除舊表
    cursor.execute('DROP TABLE IF EXISTS orders')
    cursor.execute('DROP TABLE IF EXISTS menu')

    # 建立菜單資料表 (menu)
    cursor.execute('''
        CREATE TABLE menu (
            name TEXT PRIMARY KEY,
            price INTEGER,
            stock INTEGER
        )
    ''')

    # 建立訂單資料表 (orders)
    cursor.execute('''
        CREATE TABLE orders (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            student_name TEXT,
            item_name TEXT,
            quantity INTEGER,
            status TEXT
        )
    ''')

    # 放入初始資料
    initial_menu = [
        ('美式咖啡', 50, 5),
        ('拿鐵', 65, 4),
        ('抹茶歐蕾', 70, 3)
    ]
    cursor.executemany('INSERT INTO menu (name, price, stock) VALUES (?, ?, ?)', initial_menu)

    conn.commit()
    conn.close()

# 執行初始化
init_db()
print("資料庫與初始資料建立完成！")
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 品項, 數量, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (4) AI 解讀函式
# 使用 dict 暫存已解讀過的使用者語句，避免重複呼叫 AI
# (4) AI 解讀函式（修正版）
# (4) AI 解讀函式（完全修正版）
# (4) AI 解讀函式（極致穩定版）
def parse_intent(user_input):
    user_str = user_input.strip()

    # 1. 動作關鍵字直接判斷
    action = "未知"
    if any(k in user_str for k in ["訂", "買", "要", "給我", "來"]):
        action = "訂購"
    elif any(k in user_str for k in ["取消", "退", "不要"]):
        action = "取消"
    elif any(k in user_str for k in ["查", "幾杯", "還有"]):
        action = "查詢"
    elif any(k in user_str for k in ["統計", "賣", "營收", "怎樣", "如何"]):
        action = "統計"

    # 2. 品項名稱直接判斷
    valid_items = ["美式咖啡", "拿鐵", "抹茶歐蕾"]
    item = None
    for vi in valid_items:
        if vi in user_str:
            item = vi
            break

    # 3. 數量判斷
    qty_match = re.search(r'\d+', user_str)
    if qty_match:
        qty = int(qty_match.group())
    else:
        num_map = {'一': 1, '二': 2, '兩': 2, '三': 3, '四': 4, '五': 5}
        qty = 1
        for ch, n in num_map.items():
            if ch in user_str:
                qty = n
                break

    # 如果關鍵字無法完全對應，才呼叫 LLM 輔助
    raw_response_text = ""
    if action == "未知" or (item is None and action != "統計"):
        prompt = f"""
        你是一個校園咖啡廳的語意解析助手。請從使用者的語句中萃取三個欄位：
        1. action: 動作，只能是 "訂購"、"取消"、"查詢"、"統計" 其中之一。
        2. item: 品項名稱（美式咖啡、拿鐵、抹茶歐蕾），若無則填 null。
        3. quantity: 數量，整數。若未提及數量，預設為 1。

        使用者語句："{user_str}"
        請僅回傳 JSON 格式：{{"action": "...", "item": null或品項字串, "quantity": ...}}
        """
        for attempt in range(3):
            try:
                completion = client.chat.completions.create(
                    model=MODEL,
                    messages=[{"role": "user", "content": prompt}],
                    temperature=0,
                    reasoning_effort=REASONING_EFFORT
                )
                raw_response_text = completion.choices[0].message.content.strip()
                time.sleep(2)

                clean_c = raw_response_text
                if clean_c.startswith("```"):
                    clean_c = clean_c.split("```")[1]
                    if clean_c.startswith("json"):
                        clean_c = clean_c[4:]
                data = json.loads(clean_c.strip())

                if data.get("action") in ["訂購", "取消", "查詢", "統計"]:
                    action = data.get("action")
                if data.get("item") in valid_items:
                    item = data.get("item")
                if isinstance(data.get("quantity"), int):
                    qty = data.get("quantity")
                break
            except Exception as e:
                if "429" in str(e) and attempt < 2:
                    time.sleep(30)
                else:
                    raw_response_text = str(e)

    if action == "統計":
        item = None

    return {"raw_response": raw_response_text or f"action={action}, item={item}, qty={qty}", "action": action, "item": item, "quantity": qty}

print("AI 解讀函式（極致穩定版）載入完成！")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (5) AI 解讀測試 (8句)
# 格式為 (使用者說的話, 預期動作, 預期對象/品項, 預期數量)
# "*" 表示不在意該欄位檢查
test_cases = [
    ("我要兩杯拿鐵", "訂購", "拿鐵", 2),
    ("美式咖啡還有幾杯", "查詢", "美式咖啡", 1),
    ("取消拿鐵", "取消", "拿鐵", 1),
    ("今天各品項賣得怎樣", "統計", "*", 1),
    ("給我一杯抹茶歐蕾", "訂購", "抹茶歐蕾", 1),
    ("我要買十杯美式咖啡", "訂購", "美式咖啡", 10),
    ("幫我拿一杯紅茶", "訂購", "*", 1),
    ("天氣很好想喝咖啡", "未知", "*", 1)
]

passed_count = 0
print("="*60)
print("【 AI 解讀測試開始 】")
print("="*60)

for i, (sentence, exp_action, exp_item, exp_qty) in enumerate(test_cases, 1):
    res = parse_intent(sentence)

    act_match = (res["action"] == exp_action) or (exp_action == "*")
    item_match = (res["item"] == exp_item) or (exp_item == "*")
    qty_match = (res["quantity"] == exp_qty) or (exp_qty == "*")

    is_passed = act_match and item_match and qty_match
    if is_passed:
        passed_count += 1
        status_str = "【通過】"
    else:
        status_str = "【未過】"

    print(f"測試 {i}: {status_str} 語句：「{sentence}」")
    print(f"  > AI 原始回覆: {res['raw_response'].strip()}")
    print(f"  > 系統解讀結果: 動作={res['action']}, 品項={res['item']}, 數量={res['quantity']}")
    print(f"  > 預期目標對比: 動作={exp_action}, 品項={exp_item}, 數量={exp_qty}")
    print("-" * 60)

print(f"AI 解讀測試總計執行 {len(test_cases)} 句，通過數：{passed_count} / {len(test_cases)}")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (6) 功能函式
# 處理訂購、取消、查詢與統計的核心業務邏輯（含 try/finally 與參數化查詢）
def process_order_system(user_input, name=""):
    parsed = parse_intent(user_input)
    action = parsed["action"]
    item = parsed["item"]
    qty = parsed["quantity"]

    conn = sqlite3.connect('cafe.db')
    cursor = conn.cursor()
    try:
        # 取得合法品項清單
        cursor.execute("SELECT name FROM menu")
        valid_items = [row[0] for row in cursor.fetchall()]

        if action == "未知" or (item not in valid_items and action in ["訂購", "取消", "查詢"]):
            return f"無法理解您的需求。本咖啡廳提供：美式咖啡、拿鐵、抹茶歐蕾。您可以說「我要兩杯拿鐵」、「查詢美式咖啡庫存」、「取消拿鐵」或「今天各品項賣得怎樣」。"

        # 1. 查詢功能（不需姓名）
        if action == "查詢":
            cursor.execute("SELECT stock FROM menu WHERE name = ?", (item,))
            row = cursor.fetchone()
            if row is not None:
                return f"【查詢結果】{item} 目前剩餘庫存為 {row[0]} 杯。"
            return f"找不到品項：{item}"

        # 3. 統計查詢（使用 JOIN 與 GROUP BY，無人訂的品項也要列出）
        if action == "統計" or "統計" in user_input or "賣得怎樣" in user_input or "營收" in user_input:
            query = """
                SELECT
                    m.name,
                    m.price,
                    (m.stock + COALESCE(SUM(CASE WHEN o.status = '已訂購' THEN o.quantity ELSE 0 END), 0)) AS initial_stock,
                    COALESCE(SUM(CASE WHEN o.status = '已訂購' THEN o.quantity ELSE 0 END), 0) AS sold_count,
                    m.stock AS remaining_stock,
                    (COALESCE(SUM(CASE WHEN o.status = '已訂購' THEN o.quantity ELSE 0 END), 0) * m.price) AS revenue
                FROM menu m
                LEFT JOIN orders o ON m.name = o.item_name
                GROUP BY m.name, m.price, m.stock
            """
            cursor.execute(query)
            rows = cursor.fetchall()

            result_str = "【今日銷售統計報告】\n" + "-"*40 + "\n"
            total_revenue = 0
            for r in rows:
                p_name, price, init_s, sold, remain, rev = r
                total_revenue += rev
                result_str += f"品項：{p_name}\n  單價：{price} 元\n  原始庫存：{init_s} 杯\n  已售杯數：{sold} 杯\n  剩餘庫存：{remain} 杯\n  營收：{rev} 元\n" + "-"*40 + "\n"
            result_str += f"總營收：{total_revenue} 元"
            return result_str

        # 訂購與取消動作需要填寫姓名
        if not name or not name.strip():
            return "請輸入您的姓名才能進行訂購或取消操作！"
        name = name.strip()

        # 2. 訂購功能 (A1 數量 1~3, A2 每人上限 3 杯)
        if action == "訂購":
            if not (1 <= qty <= 3):
                return f"訂購失敗：每筆訂單數量必須在 1 到 3 之間（您輸入的數量為 {qty}）。"

            # 檢查個人當日已訂購總杯數（已取消的不計）
            cursor.execute("""
                SELECT SUM(quantity) FROM orders
                WHERE student_name = ? AND status = '已訂購'
            """, (name,))
            res = cursor.fetchone()
            current_user_cups = res[0] if res and res[0] is not None else 0

            if current_user_cups + qty > 3:
                return f"訂購失敗：每位學生今天最多只能訂 3 杯，您目前已訂 {current_user_cups} 杯，無法再加購 {qty} 杯。"

            # 檢查菜單庫存
            cursor.execute("SELECT price, stock FROM menu WHERE name = ?", (item,))
            menu_row = cursor.fetchone()
            if menu_row is None:
                return f"找不到品項：{item}"

            price, stock = menu_row
            if stock < qty:
                return f"訂購失敗：{item} 庫存不足（目前剩餘 {stock} 杯，欲訂購 {qty} 杯）。"

            # 扣除庫存並新增訂單
            cursor.execute("UPDATE menu SET stock = stock - ? WHERE name = ?", (qty, item))
            cursor.execute("INSERT INTO orders (student_name, item_name, quantity, status) VALUES (?, ?, ?, ?)",
                           (name, item, qty, '已訂購'))
            conn.commit()

            total_price = price * qty
            return f"訂購成功！【{name}】訂購了 {qty} 杯 {item}，總金額共 {total_price} 元。"

        # 4. 取消功能
        if action == "取消":
            cursor.execute("""
                SELECT id, quantity FROM orders
                WHERE student_name = ? AND item_name = ? AND status = '已訂購'
                ORDER BY id DESC LIMIT 1
            """, (name, item))
            order_row = cursor.fetchone()

            if order_row is None:
                return f"取消失敗：查無【{name}】訂購「{item}」的有效訂單紀錄。"

            order_id, order_qty = order_row

            # 更新訂單狀態並釋放庫存
            cursor.execute("UPDATE orders SET status = '已取消' WHERE id = ?", (order_id,))
            cursor.execute("UPDATE menu SET stock = stock + ? WHERE name = ?", (order_qty, item))
            conn.commit()

            return f"取消成功：已將【{name}】的 {order_qty} 杯 {item} 訂單取消，並已釋放庫存。"

        return "無法辨識您的請求。"
    finally:
        conn.close()

print("功能函式載入完成！")
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (7) 主流程與流程測試 (8～10 步)
init_db() # 確保測試前重新初始化資料庫

steps = [
    ("步驟 1：查詢拿鐵庫存", "查詢拿鐵", ""),
    ("步驟 2：小明訂購 2 杯拿鐵", "我要兩杯拿鐵", "小明"),
    ("步驟 3：確認拿鐵剩餘庫存減少", "查詢拿鐵", ""),
    ("步驟 4：小華訂購 1 杯美式咖啡", "我要一杯美式咖啡", "小華"),
    ("步驟 5：小明嘗試超量訂購 3 杯美式咖啡（觸發每人3杯上限）", "我要三杯美式咖啡", "小明"),
    ("步驟 6：小明取消剛才訂購的拿鐵", "取消拿鐵", "小明"),
    ("步驟 7：再次查詢拿鐵庫存（應已釋放）", "查詢拿鐵", ""),
    ("步驟 8：小明重新訂購 1 杯抹茶歐蕾", "我要一杯抹茶歐蕾", "小明"),
    ("步驟 9：查詢今日銷售統計 (A3)", "今天各品項賣得怎樣", ""),
    ("步驟 10：隨意輸入無法辨識的語句測試容錯", "哈囉老闆在嗎", "")
]

print("="*60)
print("【 流程測試開始 (10步) 】")
print("="*60)

for title, sentence, name in steps:
    print(f"--- {title} ---")
    print(f"輸入需求：{sentence} (姓名: {name if name else '無'})")
    res = process_order_system(sentence, name)
    print(f"系統回應：\n{res}\n" + "-"*60)

print("="*60)
print("【 流程測試結束，最終資料表狀態檢查 】")
print("="*60)
show_tables()
````
